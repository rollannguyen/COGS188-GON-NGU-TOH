# Abstract

Penetration testing is typically a labor-intensive, long-winded, and often unrewarding process where cybersecurity professionals sequentially probe networks, find and exploit vulnerabilities, and gain access to secured assets. When posing a machine learning lens onto this process, we can describe automated penetration testing as a Markov decision process, with a partially observable environment, where reinforcement learning agents attack simulated enterprise networks. We propose comparing tabluar Q-learning, DGN, and PPO to analyze how perceived reward and reward design affects convergence, attack efficiency, and generalization to real-world networks, using action masking against uninformed, heuristic, and oracle search baselines to evaluate success rate, path efficiency, sample efficiency, and optimality gap.

# Problem Statement

Given a computer network that has unknown topology and vulnerabilities, the model must learn a policy to choose efficient scans, exploits, and escalation actions that allow for compromising a target host. As the agent enacts within this partially unobservable environment, it must learn the topology of the network as it goes, optimizing delayed reward (with the biggest reward as the executive device attacked) over risk within long attack chains. Specifically, we want to analyze how perceived reward and reward design affects learning speed, efficiency, and generalization of reinforcement learning agents doing automated penetration testing.

# Proposed Solution

1. Translate and formalize the process of automated penetration testing as a Partially Observable Markov Decision Process: we have an observation vector from each hosts' knowledge, and action masking for invalid and/or suboptimal actions.
2. Create baseline comparision approaches: randomized agent, rule-based greedy, breadth first search, depth first search, and A* with full network topology knowledge
3. Train RL agents: tabular Q-learning, DQN (with action masking), PPO (with action masking), ensuring that partial observability is maintained.
4. Evaluate reward design across formulations: compare potential-based shaping vs. other exploration strategies (like epsilon-greedy, UCB, count-based).
5. (Given project contraints) Tune hyperparameters on a relatively medium sized network, then evaluate on unseen networks.

# Evaluation Metrics

To measure how well our agent performs, we will test it on a large number of simulated network attacks and record the results. We will repeat this with several different random starting points (seeds) and report the average, so that one lucky or unlucky run does not skew the results. We will use the same metrics for our benchmark models (a random player and a simple rule-based player) and for our learned agents, so the comparison is fair.

**1. Success rate:** The percentage of test runs where the agent reaches the target computer within the allowed number of steps. This is our main metric because it is easy to understand and does not depend on how the reward is designed.

Success rate = (number of successful runs) / (total number of runs)

**2. Average steps to succeed:** Among the successful runs, the average number of actions the agent took. Fewer steps means a faster, more efficient attack.

Average steps = (total steps across successful runs) / (number of successful runs)

**3. Average reward:** The average total reward the agent earns per run. Reward is the score the simulator gives the agent for good actions (for example, successfully breaking into a computer) and penalties for bad ones (for example, costly actions). Because we plan to compare different reward designs, we will also re-score every agent using one fixed reward formula so that the comparison is fair.

**4. Invalid action rate:** The percentage of actions that were impossible or pointless, such as attacking a computer the agent has not discovered yet. This shows how much effort the agent wastes and lets us measure the benefit of "action masking," a technique that blocks impossible actions.

Invalid action rate = (number of invalid actions) / (total actions)

**5. Generalization:** We will train the agent on some networks and then test it on different networks it has never seen (with different numbers of computers, operating systems, and vulnerabilities). We will measure the drop in success rate between the two. A small gap means the agent learned a general strategy instead of memorizing one network.

Generalization gap = (success rate on training networks) - (success rate on new networks)

**6.Sample efficiency:** The number of training episodes an agent needs before its success rate reaches a target level (for example, 90%). Fewer episodes means the agent learns faster. This applies only to the learned agents (Q-learning, DQN, PPO), and it lets us measure how reward design affects convergence speed.

Sample efficiency = number of training episodes until success rate first reaches 90%

**7. Optimality gap:** How much longer the agent's attacks are compared with the best possible attack path, which the A* oracle finds because it can see the entire network. We compute it on successful runs only. A gap of 0 means the agent matches the oracle, and 0.25 means its attacks are 25% longer than the best possible. Because the A* oracle sees the whole network, it serves as an upper bound rather than a fair competitor. We do not expect our agents to beat it.

Optimality gap = (agent's steps - A* steps) / (A* steps)

We will consider our project successful if our learned agents (1) have a much higher success rate than the random and rule-based benchmarks, (2) complete attacks in fewer steps, and (3) keep a small generalization gap on new networks.
